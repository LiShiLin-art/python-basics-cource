## 🛠 Утилита для автопроверок

Следующая ячейка определяет вспомогательные функции. **Запустите её один раз перед началом практики.**

````python
# Утилиты для автоматической проверки заданий.
# Запустите эту ячейку один раз.


def check_value(name, actual, expected):
    """Проверяет, что переменная равна ожидаемому значению."""
    if actual == expected:
        print(f"✅ Задание пройдено: {name} = {actual!r}")
        return True
    else:
        print(f"❌ Ошибка: ожидалось {name} = {expected!r}, получено {actual!r}")
        return False


def check_function(func, cases):
    """
    Проверяет функцию на наборе тестовых случаев.
    cases — список кортежей (args, expected).
    """
    for case in cases:
        try:
            if len(case) == 2:
                args, expected = case
                actual = func(*args) if isinstance(args, tuple) else func(args)
            else:
                args, kwargs, expected = case
                actual = func(*args, **kwargs)
        except Exception as e:
            print(f"❌ Ошибка при вызове {func.__name__}: {type(e).__name__}: {e}")
            return False
        if actual != expected:
            print(f"❌ Неверный результат: {func.__name__}{args} → {actual!r}, ожидалось {expected!r}")
            return False
    print(f"✅ Функция {func.__name__} прошла все тесты")
    return True


print("Утилиты автопроверки загружены.")
````

# Часть 1. Классы и объекты

## 1.1. Зачем нужно ООП

До сих пор мы хранили данные в отдельных переменных, списках и словарях. Но когда данные и действия над ними связаны, удобнее объединить их в **объект**.

Представьте приложение-тренажёр английских слов. У каждой карточки есть:

- **данные:** слово, перевод, пример;
- **действия:** показать слово, проверить ответ, отметить «выучено».

**Объектно-ориентированное программирование (ООП)** позволяет описать такую сущность один раз — в **классе**, а потом создавать сколько угодно **объектов** этого класса.

### 1.1.1. Основные понятия

| Понятие | Что это | Аналогия |
|---------|---------|----------|
| Класс | «Чертёж», описание устройства объектов | Рецепт |
| Объект (экземпляр) | Конкретная сущность, созданная по классу | Блюдо по рецепту |
| Атрибут | Данные объекта | Ингредиенты |
| Метод | Функция, «принадлежащая» классу | Действия (нарезать, перемешать) |

Класс — это тип, а объект — конкретное значение этого типа, как `str` и `"hello"`.

## 1.2. Класс и объект

Класс объявляется ключевым словом `class`. Имя класса по PEP 8 пишется в **PascalCase** (каждое слово с заглавной буквы):

    class Flashcard:
        pass

Объект создаётся вызовом класса как функции:

    card = Flashcard()

### 1.2.1. Атрибуты

У объекта можно заводить атрибуты «на лету»:

    card.word = "cat"
    card.translation = "кот\"

````python
# Пример 1. Простейший класс и объект
class Flashcard:
    pass

card = Flashcard()
card.word = "cat"
card.translation = "кот"

print(type(card).__name__)   # Flashcard
print(card.word)             # cat
print(card.translation)      # кот
````

## 1.3. `__init__` и `self`

Задавать атрибуты вручную неудобно. Для этого есть специальный метод **`__init__`** — он вызывается автоматически при создании объекта.

### 1.3.1. `self`

`self` — это первый параметр каждого метода, ссылка на **сам объект**. Через `self` метод обращается к атрибутам и другим методам объекта.

    class Flashcard:
        def __init__(self, word, translation):
            self.word = word
            self.translation = translation

    card = Flashcard("cat", "кот")
    print(card.word)   # cat

**Разбор:**
- `__init__(self, word, translation)` — при создании передаём `word` и `translation` (без `self` — его Python подставляет сам);
- `self.word = word` — создаём атрибут `word` у конкретного объекта.

### 1.3.2. Имя `self` — соглашение

Первый параметр можно назвать как угодно, но по соглашению это всегда `self`. Менять не рекомендуется.

````python
# Пример 2. __init__ и self
class Flashcard:
    def __init__(self, word, translation):
        self.word = word
        self.translation = translation

card1 = Flashcard("cat", "кот")
card2 = Flashcard("dog", "собака")

print(card1.word, "→", card1.translation)
print(card2.word, "→", card2.translation)
````

## 1.4. Методы

**Метод** — это функция, определённая внутри класса. Методы всегда принимают `self` первым аргументом и работают с атрибутами объекта.

    class Flashcard:
        def __init__(self, word, translation):
            self.word = word
            self.translation = translation

        def show(self):
            return f"{self.word} → {self.translation}"

        def check(self, answer):
            return answer.lower().strip() == self.translation.lower().strip()

Вызов метода идёт через точку; `self` подставляется автоматически:

    card = Flashcard("cat", "кот")
    card.show()          # 'cat → кот'
    card.check("КОТ")    # True

````python
# Пример 3. Методы
class Flashcard:
    def __init__(self, word, translation):
        self.word = word
        self.translation = translation

    def show(self):
        return f"{self.word} → {self.translation}"

    def check(self, answer):
        return answer.lower().strip() == self.translation.lower().strip()

card = Flashcard("cat", "кот")
print(card.show())
print(card.check("кот"))
print(card.check("кошка"))
````

## 1.5. Атрибуты класса и атрибуты экземпляра

Атрибуты, заданные через `self`, — это атрибуты **экземпляра**: у каждого объекта они свои.

Атрибуты, объявленные прямо в теле класса, — атрибуты **класса**: они общие для всех объектов.

    class Flashcard:
        category = "English"        # атрибут класса

        def __init__(self, word, translation):
            self.word = word        # атрибут экземпляра
            self.translation = translation

    a = Flashcard("cat", "кот")
    b = Flashcard("dog", "собака")

    print(a.category, b.category)   # English English — общее значение
    print(a.word, b.word)           # cat dog — свои у каждого

**Осторожно с изменяемыми атрибутами класса:** список или словарь на уровне класса будет общим для всех объектов — это частая причина ошибок.

````python
# Пример 4. Атрибуты класса и экземпляра
class Flashcard:
    category = "English"

    def __init__(self, word, translation):
        self.word = word
        self.translation = translation

a = Flashcard("cat", "кот")
b = Flashcard("dog", "собака")

print(a.category, b.category)
print(a.word, b.word)
print(Flashcard.category)   # доступ к атрибуту класса через класс
````

# Часть 2. Инкапсуляция и наследование

## 2.1. Инкапсуляция и приватные атрибуты

**Инкапсуляция** — принцип «спрятать внутренние детали, показать только нужное». В Python нет настоящих приватных атрибутов, но есть соглашения:

| Запись | Смысл |
|--------|-------|
| `self.name` | публичный атрибут |
| `self._name` | «защищённый» — по соглашению не трогать снаружи |
| `self.__name` | «приватный» — Python искажает имя (name mangling) |

Двойное подчёркивание заставляет Python переименовать атрибут: `self.__score` → `self._Student__score`. Это защищает от случайного доступа, но **не** от намеренного.

### 2.1.1. Зачем прятать

Чтобы менять внутреннее представление, не ломая внешний код, и чтобы защитить данные от некорректных значений (например, отрицательный балл).

## 2.2. Методы доступа

Вместо прямого доступа к приватному атрибуту пишут методы-«геттеры» и «сеттеры», которые проверяют корректность значений.

    class Student:
        def __init__(self, name, score):
            self.name = name
            self.__score = None
            self.set_score(score)

        def get_score(self):
            return self.__score

        def set_score(self, value):
            if 0 <= value <= 100:
                self.__score = value
            else:
                raise ValueError("Балл должен быть от 0 до 100")

### 2.2.1. Свойства `@property`

Современный способ — декоратор `@property`, который позволяет обращаться к атрибуту как к обычному, но с проверкой:

    class Student:
        def __init__(self, score):
            self._score = score

        @property
        def score(self):
            return self._score

        @score.setter
        def score(self, value):
            if 0 <= value <= 100:
                self._score = value
            else:
                raise ValueError("Балл должен быть от 0 до 100")

    s = Student(85)
    print(s.score)    # 85
    s.score = 90      # проверка в сеттере

````python
# Пример 5. Приватный атрибут и методы доступа
class Student:
    def __init__(self, name, score):
        self.name = name
        self.__score = None
        self.set_score(score)

    def get_score(self):
        return self.__score

    def set_score(self, value):
        if 0 <= value <= 100:
            self.__score = value
        else:
            raise ValueError("Балл должен быть от 0 до 100")

s = Student("Anna", 85)
print(s.get_score())
s.set_score(90)
print(s.get_score())

# Прямой доступ к __score не сработает
try:
    print(s.__score)
except AttributeError as e:
    print("Ошибка:", e)
````

## 2.3. Наследование

**Наследование** позволяет создать класс на основе другого, унаследовав его атрибуты и методы.

    class WordCard(Flashcard):
        pass

Здесь `WordCard` — **дочерний** класс, `Flashcard` — **родительский**. Дочерний класс получает всё из родительского и может **переопределять** методы и добавлять новые.

    class WordCard(Flashcard):
        def __init__(self, word, translation, example):
            super().__init__(word, translation)   # вызов __init__ родителя
            self.example = example

        def show_full(self):
            return f"{self.show()}\nПример: {self.example}"

### 2.3.1. `super()`

`super()` — способ обратиться к методам родительского класса.

    class Child(Parent):
        def __init__(self, a, b):
            super().__init__(a)   # вызвать __init__ родителя
            self.b = b

`super().__init__(...)` — стандартный приём: сначала инициализировать часть родителя, потом добавить своё.

````python
# Пример 6. Наследование и super()
class Flashcard:
    def __init__(self, word, translation):
        self.word = word
        self.translation = translation

    def show(self):
        return f"{self.word} → {self.translation}"

class WordCard(Flashcard):
    def __init__(self, word, translation, example):
        super().__init__(word, translation)
        self.example = example

    def show_full(self):
        return f"{self.show()} | Пример: {self.example}"

card = WordCard("cat", "кот", "The cat is sleeping.")
print(card.show())       # метод родителя
print(card.show_full())  # свой метод
print(isinstance(card, Flashcard))  # True
````

## 2.4. Полиморфизм

**Полиморфизм** — способность разных классов отвечать на один и тот же вызов по-своему. Достигается переопределением методов.

    class Animal:
        def speak(self):
            return "..."

    class Cat(Animal):
        def speak(self):
            return "meow"

    class Dog(Animal):
        def speak(self):
            return "woof"

    for a in [Cat(), Dog()]:
        print(a.speak())   # meow / woof

Один и тот же код `a.speak()` работает с разными типами — это и есть полиморфизм.

````python
# Пример 7. Полиморфизм
class Animal:
    def speak(self):
        return "..."

class Cat(Animal):
    def speak(self):
        return "meow"

class Dog(Animal):
    def speak(self):
        return "woof"

for a in [Cat(), Dog(), Animal()]:
    print(type(a).__name__, "→", a.speak())
````

# Часть 3. Магические методы

## 3.1. `__str__` и `__repr__`

Магические методы (dunder-методы) — методы с двойным подчёркиванием, которые Python вызывает автоматически в определённых ситуациях.

| Метод | Когда вызывается | Цель |
|-------|------------------|------|
| `__str__` | `str(obj)`, `print(obj)` | Читаемое описание для человека |
| `__repr__` | `repr(obj)`, вывод в REPL | Однозначное описание для разработчика |

    class Flashcard:
        def __init__(self, word, translation):
            self.word = word
            self.translation = translation

        def __repr__(self):
            return f"Flashcard({self.word!r}, {self.translation!r})"

## 3.2. `__eq__`, `__lt__` — сравнения

По умолчанию `==` сравнивает объекты по ссылке (два разных объекта не равны, даже если поля одинаковые). Переопределив `__eq__`, можно сравнивать по содержимому.

## 3.4. Перегрузка операторов: `__add__`

## 3.5. Когда НЕ нужно ООП

ООП — не панацея. Не стоит создавать класс ради класса.

**Используйте функции, если:**
- задача — просто преобразование данных;
- нет устойчивого состояния, которое нужно хранить;
- класс состоял бы из одного метода.

**Используйте класс, если:**
- есть данные и связанные с ними действия;
- нужно несколько объектов с одинаковым поведением;
- нужно наследование и полиморфизм.

Простые скрипты анализа текста часто проще писать на функциях, а не на классах.

# Часть 4. Типичные ошибки

| Ошибка | Что происходит | Как исправить |
|--------|----------------|----------------|
| Забыли `self` в методе | `TypeError: f() takes 0 positional arguments but 1 was given` | Добавьте `self` первым параметром |
| Забыли `self.` при обращении к атрибуту | `NameError: name 'word' is not defined` | Пишите `self.word` |
| Изменяемый атрибут класса | Список/словарь общий для всех объектов | Заводите их в `__init__` через `self` |
| `__init__` с `return` не-`None` | `TypeError` | `__init__` не возвращает значение (кроме `None`) |
| Путаница `==` и `is` | `==` — по содержимому, `is` — по ссылке | Для равенства переопределите `__eq__`; `is` — только для `None` |
| Забыли вызвать `super().__init__` | Атрибуты родителя не инициализированы | Вызывайте `super().__init__(...)` в `__init__` дочернего класса |
| Обращение к `__private` снаружи | `AttributeError` | Используйте методы доступа или свойства |
| Определён только `__str__` | Вывод списка объектов показывает адреса | Определите `__repr__` |
