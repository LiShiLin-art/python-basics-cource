## 🛠 Утилита для автопроверок

Следующая ячейка определяет вспомогательные функции. **Запустите её один раз перед началом практики.**

````python
# Утилиты для автоматической проверки заданий.
# Запустите эту ячейку один раз.

import io
import contextlib


def check_value(name, actual, expected):
    """Проверяет, что переменная равна ожидаемому значению."""
    if actual == expected:
        print(f"✅ Задание пройдено: {name} = {actual!r}")
        return True
    else:
        print(f"❌ Ошибка: ожидалось {name} = {expected!r}, получено {actual!r}")
        return False


def check_function(func, cases):
    """
    Проверяет функцию на наборе тестовых случаев.
    cases — список кортежей (args, expected) или (args, kwargs, expected).
    """
    for case in cases:
        try:
            if len(case) == 2:
                args, expected = case
                actual = func(*args) if isinstance(args, tuple) else func(args)
            else:
                args, kwargs, expected = case
                actual = func(*args, **kwargs)
        except Exception as e:
            print(f"❌ Ошибка при вызове {func.__name__}: {type(e).__name__}: {e}")
            return False
        if actual != expected:
            print(f"❌ Неверный результат: {func.__name__}{args} → {actual!r}, ожидалось {expected!r}")
            return False
    print(f"✅ Функция {func.__name__} прошла все тесты")
    return True


print("Утилиты автопроверки загружены.")
````

# Часть 1. Определение функций

## 1.1. Зачем нужны функции

Представьте, что вам нужно **три раза** подсчитать среднее значение списка оценок в одной программе. Без функций пришлось бы писать один и тот же код три раза. Это плохо:

- **дублирование** — код разрастается;
- **ошибки** — при исправлении легко что-то забыть в одном из копий;
- **читаемость** — непонятно, что делает блок;
- **тестирование** — сложно проверить каждую копию отдельно.

**Функция** — это именованный блок кода, который можно вызвать по имени сколько угодно раз, передать ему данные (аргументы) и получить результат (возвращаемое значение).

### 1.1.1. Что уже знакомо: встроенные функции

Вы уже пользовались функциями из стандартной библиотеки:

    len("hello")        # 5
    max([1, 5, 3])      # 5
    print("hello")      # вывод на экран

Теперь научимся создавать **свои** функции.

## 1.2. Синтаксис `def`

### 1.2.1. Простейшая функция

    def hello():
        print("Hello, world!")

    hello()   # вызов функции

**Разбор:**

| Часть | Что означает |
|-------|--------------|
| `def` | Ключевое слово (от англ. *define* — определять) |
| `hello` | Имя функции (по PEP 8 — в нижнем регистре, через `_`) |
| `()` | Параметры (пока пустые) |
| `:` | Двоеточие, как у `if` и `for` |
| Тело с отступом | Всё, что выполняется при вызове |

### 1.2.2. Функция с параметром

    def hello(name):
        print(f"Hello, {name}!")

    hello("Anna")    # Hello, Anna!
    hello("Ivan")    # Hello, Ivan!

Здесь `name` — **параметр** функции, `"Anna"` — **аргумент** при вызове.

### 1.2.3. Правила именования

- только буквы, цифры, подчёркивания (не начинается с цифры);
- регистрозависимо: `Hello` и `hello` — разные функции;
- по PEP 8 — `snake_case`: `count_words`, `normalize_text`, `is_palindrome`;
- не используйте имена встроенных функций (`print`, `len`, `max` и т. п.) — вы их перезапишете.

````python
# camel_case: countWords, normalizeText, isPalindrome
# Пример 1. Простейшие функции
def hello():
    print("Hello, world!")

def hello_name(name):
    print(f"Hello, {name}!")

def hello_name_r(name) -> str:
    return f"Добрый день, {name}!"



hello()
name = "Anna"
hello_name(name)
n = "Ivan"
hello_name(name)
m = hello_name_r(n)
print(m)
````

## 1.3. `return` — возврат значения

Функция может не только печатать, но и **возвращать** результат. Возвращаемое значение можно сохранить в переменную, передать в другую функцию, использовать в вычислениях.

### 1.3.1. Синтаксис

    def square(x):
        return x * x

    result = square(5)
    print(result)   # 25

### 1.3.2. Разница между `print` и `return`

| `print` | `return` |
|---------|----------|
| Показывает значение на экране | Возвращает значение из функции |
| Ничего не возвращает (возвращает `None`) | Возвращает то, что указано |
| Нельзя сохранить в переменную | Можно сохранить и использовать |

**Пример разницы:**

    def f_print(x):
        print(x * x)

    def f_return(x):
        return x * x

    a = f_print(5)     # напечатает 25, a = None
    b = f_return(5)    # ничего не печатает, b = 25

### 1.3.3. Когда функция достигает `return` — она завершается

    def check(x):
        if x > 0:
            return "positive"
        return "not positive"
        # код ниже НЕ выполнится

### 1.3.4. Функция без `return`

Если в функции нет `return`, она возвращает `None`:

    def f():
        x = 5

    print(f())   # None

### 1.3.5. Несколько `return`

В одной функции может быть несколько `return` — но выполнится только первый достигнутый:

    def classify(score):
        if score >= 90:
            return "отлично"
        elif score >= 75:
            return "хорошо"
        else:
            return "удовлетворительно"

Это — стандартная идиома Python: `return` внутри `if/elif/else`.

````python
# Пример 2. return vs print
def f_print(x):
    print(f"print: {x * x}")

def f_return(x):
    return x * x

a = f_print(5)
print(f"a = {a}")   # a = None

b = f_return(5)
print(f"b = {b}")   # b = 25
````

````python
# Пример 3. Несколько return
def classify(score):
    if score >= 90:
        return "отлично"
    elif score >= 75:
        return "хорошо"
    else:
        return "удовлетворительно"

print(classify(95))
print(classify(80))
print(classify(60))
````

## 1.4. Области видимости

### 1.4.1. Локальные переменные

Переменная, созданная **внутри** функции, называется **локальной**. Она существует только внутри этой функции.

    def f():
        x = 10    # локальная переменная
        return x

    f()
    print(x)      # NameError: name 'x' is not defined

### 1.4.2. Глобальные переменные

Переменная, созданная **вне** функции, называется **глобальной**. Она видна внутри функции, но изменять её изнутри — плохая практика.

    x = 10

    def f():
        return x    # читает глобальную

    print(f())      # 10

Чтобы изменить глобальную переменную изнутри функции, нужен `global`, но **использовать его почти никогда не стоит** — это делает код непредсказуемым.

### 1.4.3. Правило

Функция должна **получать данные через параметры** и **возвращать результат через `return`**. Не полагайтесь на глобальные переменные.

````python
# Пример 4. Области видимости
x = 10   # глобальная

def read_global():
    return x     # читаем глобальную

def write_local():
    x = 20       # это новая локальная переменная
    return x

print("Глобальная x:", x)
print("read_global():", read_global())
print("write_local():", write_local())
print("После вызова write_local глобальная x всё ещё:", x)
````

# Часть 2. Аргументы и возврат значений

## 2.1. Виды аргументов

В Python есть несколько способов передать данные в функцию. Разберём каждый.

### 2.1.1. Позиционные аргументы

Самый простой способ: значения передаются **в том же порядке**, в каком заданы параметры.

    def greet(name, level):
        return f"{name}, level {level}"

    greet("Anna", "B2")     # по порядку

### 2.1.2. Именованные (keyword) аргументы

Можно передавать по имени — тогда порядок не важен:

    greet(level="B2", name="Anna")    # тот же результат

Это удобно, когда параметров много или они легко путаются.

### 2.1.3. Аргументы по умолчанию

Если параметр может отсутствовать — задайте значение по умолчанию:

    def greet(name, level="A1"):
        return f"{name}, level {level}"

    greet("Anna")           # level = "A1" по умолчанию
    greet("Ivan", "B2")    # level = "B2"

**Правило:** параметры **без** значения по умолчанию должны идти **до** параметров **со** значением по умолчанию.

    def f(a, b=1):     # ✅ правильно
    def f(a=1, b):     # ❌ SyntaxError

### 2.1.4. Аргументы `*args` — переменное число

Если заранее неизвестно, сколько аргументов будет передано, используйте `*args`. Внутри функции это **кортеж**.

    def sum_all(*args):
        return sum(args)

    sum_all(1, 2, 3)        # 6
    sum_all(10)             # 10
    sum_all()               # 0

### 2.1.5. Аргументы `**kwargs` — именованные с произвольными именами

`**kwargs` — это словарь произвольных именованных аргументов:

    def show_info(**kwargs):
        for key, value in kwargs.items():
            print(f"{key}: {value}")

    show_info(name="Anna", age=25, level="B2")

### 2.1.6. Порядок параметров в определении

    def f(a, b=1, *args, **kwargs):
        ...

Порядок фиксирован: **позиционные → по умолчанию → `*args` → `**kwargs`**.

````python
# Пример 5. Виды аргументов
def greet(name, level="A1"):
    return f"{name}, level {level}"

print(greet("Anna"))                  # по умолчанию
print(greet("Ivan", "B2"))            # позиционно
print(greet(level="C1", name="Olga"))  # именованно

def sum_all(*args):
    return sum(args)

print("sum_all(1, 2, 3) =", sum_all(1, 2, 3))
print("sum_all(10) =", sum_all(10))
print("sum_all() =", sum_all())

def show_info(**kwargs):
    for k, v in kwargs.items():
        print(f"  {k}: {v}")

show_info(name="Anna", age=25, level="B2")
````

## 2.2. Возврат нескольких значений

Функция может вернуть несколько значений через **кортеж**:

    def min_max(lst):
        return min(lst), max(lst)

    lo, hi = min_max([3, 1, 4, 1, 5])
    print(lo, hi)     # 1 5

Технически возвращается **один** кортеж — просто его можно распаковать в переменные.

### 2.2.1. Возврат словаря

Если значений много или их нужно именовать, удобнее вернуть словарь:

    def stats(lst):
        return {
            "min": min(lst),
            "max": max(lst),
            "avg": sum(lst) / len(lst),
        }

    s = stats([3, 1, 4, 1, 5])
    print(s["avg"])   # 2.8

````python
# Пример 6. Возврат нескольких значений
def min_max(lst):
    return min(lst), max(lst)

lo, hi = min_max([3, 1, 4, 1, 5])
print(f"min = {lo}, max = {hi}")

def stats(lst):
    return {
        "min": min(lst),
        "max": max(lst),
        "avg": sum(lst) / len(lst),
    }

s = stats([3, 1, 4, 1, 5])
print("stats:", s)
````

# Часть 3. Модули и стандартная библиотека

## 3.1. Что такое модуль

**Модуль** — это файл `.py`, содержащий функции, классы и переменные. Python поставляется с большой **стандартной библиотекой** — множеством готовых модулей.

### 3.1.1. Способы импорта

    import math                    # импорт всего модуля
    math.sqrt(16)                  # доступ через точку

    from math import sqrt          # импорт конкретной функции
    sqrt(16)                       # без префикса

    import math as m               # псевдоним

# Часть 4. Типичные ошибки

| Ошибка | Что происходит | Как исправить |
|--------|----------------|----------------|
| `print` вместо `return` | Функция возвращает `None` | Замените `print(...)` на `return ...` |
| Забыли вызвать функцию | `result = square` вместо `square(5)` | Добавьте скобки |
| Изменяемое значение по умолчанию `def f(lst=[])` | Список сохраняется между вызовами | Используйте `def f(lst=None)` и потом `if lst is None: lst = []` |
| Параметр без значения после параметра со значением | `def f(a=1, b)` → `SyntaxError` | Поменяйте порядок: `def f(a, b=1)` |
| Обращение к локальной переменной вне функции | `NameError` | Возвращайте значение через `return` |
| `from module import *` | Загрязнение пространства имён | Импортируйте нужные имена явно |
| Забыли перезагрузить модуль в Colab | Используется старая версия | `importlib.reload(module)` |
| Путаница `args` и `*args` | `args` — кортеж, `*args` — упаковка | В определении `*args`, в теле `args` |
| `return` в цикле вместо после | Функция завершается на первой итерации | Выносите `return` за цикл, если нужен накопитель |
| Изменение списка внутри функции без `return` | Изменения могут происходить (мутабельность), но лучше возвращать новый | Возвращайте новый список |
